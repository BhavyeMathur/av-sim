```python
import ev_sim

RUN_DB = "runs/runs.sqlite3"
```

```python
experiment = "ann-arbor-strategy-speed-comparison"
```

```python
configs = ev_sim.generate_runs_from_yaml(f"experiments/{experiment}.yaml")
ev_sim.build_and_run("sim", f"runs/{experiment}/manifest.txt")
c = ev_sim.compute_pending_stats(RUN_DB, ev_sim.compute_stats)
print(f"completed {c} runs")
```

```python
c = ev_sim.recompute_experiment_stats(RUN_DB, experiment, ev_sim.compute_stats)
print(f"completed {c} runs")
```

```python
db = ev_sim.RunDB(RUN_DB)

df = (db.query()
      .experiment(experiment)
      .where_config("sim.length_s", "=", 24 * 3600 * 7)
      .where_config("fleet.fleet_size", "=", 15000)
      .where_config("policy.matching", "=", "greedy-h3")
      .load())

df
```
